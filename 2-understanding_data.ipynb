```python
import pandas as pd
df = pd.read_csv("files/batch-3.csv")

```

```python

print("COLUMN NAMES:")
print(df.columns.tolist())
```

```python

print("\nFIRST 5 ROWS:")
print(df.head(),end="")
```

```python

print("\nDATA TYPES:")
print(df.dtypes)
```

```python

print("\nMISSING VALUES:")
print(df.isnull().sum())
```

```python
print("\nDUPLICATE COMPLETE ROWS:")
print(df.duplicated().sum())
```

```python

print("\nDUPLICATE POST IDs:")
print(df["id"].duplicated().sum())
```

```python

print("\nUNIQUE AUTHORS:")
print(df["screen_name"].nunique())
```

```python

print("\nTOP 20 AUTHORS BY NUMBER OF POSTS:")
print(df["screen_name"].value_counts().head(20))
```

```python

print("\nTOP 20 REPEATED TEXTS:")
print(df["full_text"].value_counts().head(20))
```

```python

df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

print("\nEARLIEST POST:")
print(df["created_at"].min())

print("\nLATEST POST:")
print(df["created_at"].max())
```

```python

print("\nPOSTS PER DAY:")
print(df["created_at"].dt.date.value_counts().sort_index())
```

```python

print("\nNUMERIC COLUMN SUMMARY:")
print(df.describe())
```
